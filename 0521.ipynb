# Dictionary
## Энэ төрлийн элементийн цуглуулга нь дараалалгүй, өөрчилж болно, индекслэгдсэн байна

## ялгаатай тал нь түлхүүрээ өөрсдөө тодорхойлж өгдөг, Key-ээрээ хандаж ажиллана.

```python
# Create a dictionary
my_dict = {
   '1' : 'Даваа гараг',
   '2' : 'Мягмар гараг',
   '3' : 'Лхагва гараг','4' : 'Пүрэв гараг','5' : 'Баасан гараг',
   '6' : 'Бямба гараг', '7' : 'Ням гараг'
}
print(my_dict)
```

## Accessing items

```python
# key
print(my_dict['3'])
```

```python
# get
print(my_dict.get('10',"Таны оруулсан утга байхгүй байна. Та 1-7 ын хооронд утга өгнө үү"))
```

## Change value

```python
my_dict['1'] = 'Monday'
print(my_dict)
```

Loop through a dictionary

```python
# key eer ni
for guiltiin_utga in my_dict.keys():
    print(guiltiin_utga)
```

```python
# value aar ni
for x in my_dict.values():
    print(x)
```

```python
# items
for x, y in my_dict.items():
    print(x, y)
```

```python
# tuple хэлбэрээр хэвлэх
for d in my_dict.items():
    print(d)
```

## Заавал код зогсох нөхцөлтэй байна.

#Pass

```python
'''
Түр хоосон үлдээх команд, одоохондоо юу ч хийхгүй гэсэн утгатай. Кодонд алдаа гархаас сэргийлээд хэрэглэнэ.
'''
for i in range(1,6):
  if i ==3:
    pass
  else:
    print(i)
```

```python
'''
break  давталтыг шууд зогсооно
loop цааш үргэлжлэхгүй
'''
for i in range(1,11):
  if i == 5:
    break
  print(i)
```

```python
# Continue Одоогийн давталтыг алгасна. Дараагийн давталт руу орно
for i in range(1,11):
  if i == 5:
    continue
  print(i)
```

```python
# else
# Давталт дуусаад ашиглана
for i in range(1,6):
  print(i)
else:
  print('Давталт амжилттай дууслаа')
```
